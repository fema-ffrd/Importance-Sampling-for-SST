```python
df = pandas.read_parquet("./data/1_interim/hms_timeseries.pq")
```

```python
df["event_id"].max()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Select column
data = df["value"].dropna().sort_values()  # ascending

n = len(data)
ranks = np.arange(1, n + 1)

# Non-exceedance probability
non_exceedance = ranks / (n + 1)

# Plot
plt.figure()
plt.plot(non_exceedance, data)
plt.yscale("log")  # log Y-axis
plt.xlabel("Non-Exceedance Probability")
plt.ylabel("Flow")
plt.title("Non-Exceedance Curve")
plt.grid(True, which="both")
plt.show()
```
